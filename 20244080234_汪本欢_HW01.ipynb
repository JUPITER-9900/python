## 欧几里得算法

用辗转相除求两个整数的最大公约数。

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
# 如果出现负数
print(gcd(-12, 18))
print(gcd(-12, -18))
```

上面没加绝对值。负数时循环照样能跑完，但结果可能带负号（比如 `gcd(-12, -18)` 返回 -6），而数学上最大公约数约定为正数。所以更严谨的做法是最后 `return abs(m)`。

```python
def gcd2(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m)

print(gcd2(-12, 18))
print(gcd2(-12, -18))
```

## 近似欧拉数 e

e 的定义：$e = \sum_{k=0}^{\infty} \frac{1}{k!}$

伯努利极限：$(1+\frac{1}{x})^{x}$，当 $x \to \infty$

```python
# 1. 用极限近似，取 x = n
def euler_limit(n):
    return (1 + 1 / n) ** n

print(euler_limit(1000))
print(euler_limit(1000000))
print(euler_limit(10 ** 7))
```

```python
# 2. 用前 n 项和近似（注意 k 从 0 开始）
def euler_infinite_sum(n):
    total = 0.0
    fact = 1.0
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

print(euler_infinite_sum(0))   # 0
print(euler_infinite_sum(1))   # 只有 k=0 一项，1
print(euler_infinite_sum(4))   # 1 + 1 + 1/2 + 1/6 ≈ 2.667
```

```python
# 3. 累加和直到与真实 e 的差距小于 eps
import math

def euler_approx(eps):
    total = 0.0
    fact = 1.0
    k = 0
    while math.exp(1) - total > eps:
        if k > 0:
            fact *= k
        total += 1 / fact
        k += 1
    return total

print(euler_approx(1e-10))
print(math.exp(1))
```

```python
# 4. 打印前 n 项的求和表
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(k, euler_infinite_sum(k))

print_euler_sum_table(8)
```

## 测试整数属性

正整数 n 是 2 的幂，如果 $n = 2^{p}$ 对某个 $p \geq 0$ 成立。要求只用除法和取模，不用 `math.sqrt`。

```python
def is_power_of_2(n):
    if n <= 0:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

for x in (1, 2, 4, 16, 12, 18, 0):
    print(x, is_power_of_2(x))
```

```python
# 推广：n 是否是 b 的幂
def is_power(b, n):
    if n <= 0:
        return False
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print(is_power(2, 16))
print(is_power(3, 27))
print(is_power(2, 12))
print(is_power(5, 125))
print(is_power(2, 1))   # 1 = 2^0
```
