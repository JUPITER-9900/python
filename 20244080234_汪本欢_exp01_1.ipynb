## 实验2：函数、条件、循环与递归

### 2.1 分数等级

```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在0~100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(85))
print(score_level(72))
print(score_level(45))
try:
    score_level(120)
except ValueError as e:
    print("异常:", e)
```

### 2.2 for 和 while 求 1~100 之和

两种写法结果应该一致。

```python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for:", total_for)
print("while:", total_while)
```

### 2.3 阶乘：递归版和迭代版

约定 0! = 1，输入非负整数。

```python
def fact_rec(n):
    if n == 0:
        return 1
    return n * fact_rec(n - 1)

def fact_iter(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

for n in (0, 1, 5, 10):
    print(n, fact_rec(n), fact_iter(n))
```

### 2.4 summary(values)

返回最小值、最大值、平均值的元组，用元组解包接收。

```python
def summary(values):
    return min(values), max(values), sum(values) / len(values)

data = [3, 1, 4, 1, 5]
mn, mx, avg = summary(data)
print("min:", mn, "max:", mx, "avg:", avg)
```

### 2.5 sum_all(*args)

接收任意数量参数，返回它们的和。

```python
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1, 2, 3, 4))
```

## 实验4：文件读写、异常处理与模块

```python
import os

print("当前目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录:", data_dir)
```

创建 `students.txt`，每行 `姓名,成绩`，含几条无效记录用于后面的异常处理。

### 4.3 parse_student

把一行解析成 (姓名, 成绩)，字段数量不对、姓名为空、成绩超范围都抛 `ValueError`。

```python
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str)
    if score < 0 or score > 100:
        raise ValueError("成绩超出0~100范围")
    return name, score

# 简单测试
print(parse_student("Alice,85"))
try:
    parse_student("Grace,abc")
except ValueError as e:
    print("异常:", e)
```

### 4.4 逐行处理，捕获无效记录

跳过表头，记录出错的行号和原因，继续处理后面的记录。

```python
students = []
errors = []
with open(os.path.join(data_dir, "students.txt"), "r", encoding="utf-8") as f:
    lines = f.readlines()

# 第 1 行是表头，从第 2 行开始（行号从 2 计）
for line_no, line in enumerate(lines[1:], start=2):
    try:
        name, score = parse_student(line)
        students.append({"name": name, "score": score})
    except ValueError as e:
        errors.append((line_no, line.strip(), str(e)))

print("有效记录：", students)
print("无效记录：")
for line_no, raw, reason in errors:
    print(f"  第{line_no}行 {raw!r} -> {reason}")
```

### 4.5 写 report.txt，演示 w 和 a 模式

`w` 覆盖写入，`a` 追加到文件末尾。

```python
valid_count = len(students)
invalid_count = len(errors)
avg = sum(s["score"] for s in students) / valid_count
pass_rate = len([s for s in students if s["score"] >= 60]) / valid_count

report = (f"有效记录数：{valid_count}\n"
          f"无效记录数：{invalid_count}\n"
          f"平均分：{avg:.2f}\n"
          f"及格率：{pass_rate:.1%}\n")

report_path = os.path.join(data_dir, "report.txt")

# w 模式：覆盖已有内容
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report)
print("--- w 写入后 ---")
print(open(report_path, encoding="utf-8").read())

# a 模式：追加
with open(report_path, "a", encoding="utf-8") as f:
    f.write("这是用 a 模式追加的一行。\n")
print("--- a 追加后 ---")
print(open(report_path, encoding="utf-8").read())
```

## 实验5：类、继承与对象复制

### 5.1 定义 Student 类

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"{self.name}({self.score}分)"

    def __repr__(self):
        return self.__str__()
```

### 5.2 为每条有效记录创建对象

```python
objs = [Student(s["name"], s["score"]) for s in students]
for o in objs:
    print(o, "及格" if o.is_passed() else "不及格")
```

### 5.3 继承：GraduateStudent

增加 `research_topic`，用 `super().__init__()` 初始化父类属性，重写 `__str__()`。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"{super().__str__()} 研究方向：{self.research_topic}"

g = GraduateStudent("Bob", 58, "数据挖掘")
print(g)
```

### 5.4 直接赋值 / copy.copy / copy.deepcopy

每次比较都从新建的原始对象开始，看修改内部列表后原对象有没有被影响。

```python
import copy

# 直接赋值：同一个对象
original = {"name": "Alice", "scores": [85, 90]}
a1 = original
a1["scores"].append(100)
print("直接赋值后原对象:", original)

# 浅拷贝：外层是新对象，内部列表还是同一个
original = {"name": "Alice", "scores": [85, 90]}
b1 = copy.copy(original)
b1["scores"].append(100)
print("浅拷贝后原对象:", original)
print("浅拷贝副本:", b1)

# 深拷贝：内部列表也是新对象
original = {"name": "Alice", "scores": [85, 90]}
c1 = copy.deepcopy(original)
c1["scores"].append(100)
print("深拷贝后原对象:", original)
print("深拷贝副本:", c1)
```
