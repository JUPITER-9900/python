## 字符串的乐趣

1. `is_palindrome`：忽略空格和大小写判断回文
2. `is_abecedarian`：字母从左到右按字母序排列（允许重复）
3. `remove_vowels`：去掉 a、e、i、o、u

```python
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print(is_palindrome("level"))
print(is_palindrome("tacocat"))
print(is_palindrome("TACO cat"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("hello"))
```

```python
def is_abecedarian(s):
    s = s.replace(" ", "")
    for i in range(len(s) - 1):
        if s[i] > s[i + 1]:
            return False
    return True

print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
```

```python
def remove_vowels(s):
    return "".join(c for c in s if c not in "aeiouAEIOU")

print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
print(remove_vowels("HELLO"))
```

## 列表的乐趣

1. `list_reverse`：返回反转后的列表，输入不是列表要报错
2. `is_sorted`：只遍历一次判断是否升序

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

print(list_reverse([1, 2, 3]))
print(list_reverse([]))
try:
    list_reverse("abc")
except TypeError as e:
    print("异常:", e)
```

```python
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print(is_sorted([1, 2, 3]))
print(is_sorted([3, 1, 2]))
print(is_sorted([1, 1, 2]))
print(is_sorted((1, 2, 3)))
print(is_sorted("abc"))
```

## 作为向量的元组

用长度为 d 的元组表示 d 维向量，用元组的元组表示矩阵。

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, -2), 0.5))
try:
    vec_scalar_mult([1, 2], 2)
except TypeError as e:
    print("异常:", e)
```

```python
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度不一致")
    return float(sum(a * b for a, b in zip(x, y)))

print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1,), (2,)))
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("异常:", e)
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    # 每个元素都必须是元组
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    for row in mx:
        # 所有内层元组长度相等
        if len(row) != n:
            return False
        # 每个元素都是整数或浮点数
        if not all(isinstance(x, (int, float)) for x in row):
            return False
    return True

print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, "a"), (3, 4))))
print(check_valid_mx([[1, 2], [3, 4]]))
print(check_valid_mx(((1.5, 2), (3, 4.2))))
```

```python
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("输入不是有效矩阵")
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度不一致")
    return tuple(sum(row[i] * v[i] for i in range(len(v))) for row in mx)

print(mx_vec_mult(((1, 2), (3, 4)), (1, 1)))
print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (1, 0, 1)))
try:
    mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3))
except ValueError as e:
    print("异常:", e)
```

## 更多向量的乐趣（面向对象）

用类把向量封装起来，支持比较、点积和乘法。

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        # 没给元素时默认全 0
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("元素个数与维度不一致")
        self.dim = dim
        self.elements = tuple(elements)

    def _check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和 Vector 比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check(other)
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("维度不一致")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致")
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        if isinstance(other, (int, float)):
            return Vector(self.dim, [a * other for a in self.elements])
        raise TypeError("不支持与该类型相乘")

    def __rmul__(self, other):
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
```

维度（`dim`）和元素（`elements`）放在实例属性里：每个向量对象都有自己的维度和元素，是它自己的状态。如果所有向量都共享同一套维度和元素才适合用类属性，这里显然不是。

```python
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print("2*v =", 2 * v)
print("v*2 =", v * 2)
print("v*w =", v * w)
print("v == Vector(2,(2,3)):", v == Vector(2, (2, 3)))
print("Vector(2,(2,4)) < Vector(2,(2,5)):", Vector(2, (2, 4)) < Vector(2, (2, 5)))
print("Vector(2,(1,9)) > Vector(2,(2,0)):", Vector(2, (1, 9)) > Vector(2, (2, 0)))
print("v.dot(w) =", v.dot(w))

# 默认元素
z = Vector(3)
print("默认向量:", z)

# 维度不一致报错
try:
    Vector(2, (1, 2)) < Vector(3, (1, 2, 3))
except ValueError as e:
    print("异常:", e)
try:
    Vector(2, (1, 2)) * Vector(3, (1, 2, 3))
except ValueError as e:
    print("异常:", e)
```
